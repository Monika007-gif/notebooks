# OBJETOS MUTABLES. El contenido se puede cambiar después de haberlo creado, sin necesidad de crear un objeto nuevo.

# Listas. En Python son mutables



```python
L = [2, 3, 4]
M = L
L[0] = 24
print("L = ", L)
print("M = ", M)

```

# Objeto mutable, hace referencia al mismo objeto. Hay un cambio en el propio objeto. Podemos cambiar uno de sus elementos. La lista pasa a ser 24,3,4.L[0]=24. El 0 es el primer número


```python
L = [24, 3, 4]
M = [2, 3, 4]
```

# Modifica el objeto compartido


```python
x = 1
l = [1, 2]
changer(x, l)
print("l = ", l)
assert x == 1, "x es 1"
```

# Pasa una copia para que nuestra l [1, 2] se mantenga y no cambie. Vemos las invariantes abajo

```python
l = [1, 2]
def changer(a, b):
    b = b[:]
    a = 2
    b[0] = 'spam'
    changer(x, l)
    assert l == [1, 2]
    print("l = ", l)
    l = [1, 2]
```

# l es una invariante. Podemos convertir el objeto mutable por uno inmutable


```python
l = [1, 2]
def changer(a, b):
                a = 2
                b[0] = 'spam'
changer(x, tuple(l))
print("l = ", l)
assert l == [1,2]
```

# Modos especiales para pasar argumentos en Python. Cómo llamamos a una función, pasando información de distintas maneras. Por defecto

```python
def f(a, b, c):
    print(a, b, c)
    f(1, 2, 3)
```

# obtenemos un texto sin formato, a, b y c son los parámetros de la función, son los que usaremos para recibir valores. Esa función la llamamos f y le pasamos los tres valores, hay un orden en Python, a vale 1, b vale 2 y c vale 3, en este ejemplo.

```python
def f(a, b=2):
    print(a, b)
    f(1)
```

# A la b le damos valor predeterminado, es el que Python usará sino le damos otro.

```python
def f(a, b, c):
    print(a, b, c)
    f(a=1, b=2, c=3)
```

# Keywords, al llamar a una función puedo indicar el nombre de cada parámetro para decirle qué valor debe recibir. Hemos definido la función, a,b,c. Python asigna los valores por sus nombres, el resultado es 1,2,3. Por nombre Keywords lo que importa es el nombre del parámetro no la posición

```python
def f(a, b=2, c=3):
    print(a, b, c)
    f(1)
```
